Create descriptors

```python
import numpy as np
from rdkit import Chem
from rdkit.Chem import Descriptors

mols = meta['smiles_can'].apply(Chem.MolFromSmiles)

X= pd.DataFrame([Descriptors.CalcMolDescriptors(m) for m in mols])
X = X.replace([np.inf, -np.inf], np.nan)

X = X.loc[:, X.isna().mean() < 0.05]
X = X.loc[:,X.nunique()>1]
print(X.shape)
```

```python
X.to_parquet(f"{PROC}/descriptors.parquet")         # или X.to_csv(...)
```
